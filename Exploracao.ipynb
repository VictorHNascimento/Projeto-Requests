```python
import requests
```

```python
#Importando a chave
with open("chave/chave.txt", "r", encoding="utf-8") as f:
    ac_token = f.read()

#ac_token = 'INSIRA A CHAVE AQUI'
```

```python
# especificando a versão da API
headers = {
    'Authorization': 'Bearer' + ac_token ,
    'X-GitHub-Api-Version': '2022-11-28', 
    'Accept': 'application/vnd.github+json'}
```

```python
api_url_base = 'https://api.github.com'
owner = 'amzn'
url = f'{api_url_base}/users/{owner}/repos'
```

```python
url
```

```python
r = requests.get(url, headers=headers)
r.status_code
```

```python
r.json()
```

```python
len(r.json())
```

```python
repos_list = []
for page_n in range(1,7):
    try:
        url_page = f'{url}?page={page_n}'
        response = requests.get(url_page, headers=headers)
        repos_list.append(response.json())
    except:
        repos_list.append(None)
```

```python
repos_list
```

```python
repos_name = []
for page in repos_list:
    for repo in page:
        repos_name.append(repo['name'])
```

```python
repos_name
```

```python
repos_language = []
for page in repos_list:
    for repo in page:
        repos_language.append(repo['language'])
```

```python
repos_language
```

```python
import pandas as pd
```

```python
dados_amz = pd.DataFrame()
dados_amz['repository_name'] = repos_name
dados_amz['language'] = repos_language
```

```python
dados_amz.head()
```

```python
dados_amz.to_csv('amazon.csv')
```
